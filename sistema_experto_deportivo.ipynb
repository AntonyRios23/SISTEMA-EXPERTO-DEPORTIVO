```python
# ------------------------------------------------------------
# 2. OBTENER DATOS DEL JUGADOR Y GENERAR LOS HECHOS
# ------------------------------------------------------------

def pedir_valor(nombre):
    """
    Solicita una valoración entre 0 y 10.
    No permite letras, números negativos ni valores mayores a 10.
    """

    while True:
        try:
            valor = float(input(f"{nombre} (0-10): "))

            if 0 <= valor <= 10:
                return valor
            else:
                print(" Ingrese un valor entre 0 y 10.")

        except ValueError:
            print(" Entrada inválida. Ingrese solamente números.")


def obtener_hechos():

    print("\n" + "=" * 55)
    print("        SISTEMA EXPERTO DE FÚTBOL ")
    print("   Recomendación de posición para un jugador")
    print("=" * 55)

    print("\nIngrese los datos del jugador:")

    nombre = input("\nNombre del jugador: ").strip()

    # --------------------------------------------------------
    # Ingreso de características
    # --------------------------------------------------------

    velocidad = pedir_valor(" Velocidad")
    resistencia = pedir_valor(" Resistencia")
    pase = pedir_valor(" Pase")
    vision = pedir_valor(" Visión de juego")
    regate = pedir_valor(" Regate")
    remate = pedir_valor(" Remate")
    defensa = pedir_valor(" Defensa")
    fuerza = pedir_valor(" Fuerza")
    reflejos = pedir_valor(" Reflejos")
    juego_aereo = pedir_valor(" Juego aéreo")

    # --------------------------------------------------------
    # Convertimos las puntuaciones en HECHOS
    # --------------------------------------------------------

    hechos = set()

    # VELOCIDAD
    if velocidad >= 8:
        hechos.add("velocidad_alta")
        hechos.add("velocidad_suficiente")

    elif velocidad >= 6:
        hechos.add("velocidad_suficiente")

    else:
        hechos.add("velocidad_baja")


    # RESISTENCIA
    if resistencia >= 8:
        hechos.add("resistencia_alta")
        hechos.add("resistencia_suficiente")

    elif resistencia >= 6:
        hechos.add("resistencia_suficiente")

    else:
        hechos.add("resistencia_baja")


    # PASE
    if pase >= 8:
        hechos.add("pase_alto")
        hechos.add("pase_suficiente")

    elif pase >= 6:
        hechos.add("pase_suficiente")

    else:
        hechos.add("pase_bajo")


    # VISIÓN DE JUEGO
    if vision >= 8:
        hechos.add("vision_alta")

    elif vision >= 6:
        hechos.add("vision_media")

    else:
        hechos.add("vision_baja")


    # REGATE
    if regate >= 8:
        hechos.add("regate_alto")

    elif regate >= 6:
        hechos.add("regate_medio")

    else:
        hechos.add("regate_bajo")


    # REMATE
    if remate >= 8:
        hechos.add("remate_alto")

    elif remate >= 6:
        hechos.add("remate_medio")

    else:
        hechos.add("remate_bajo")


    # DEFENSA
    if defensa >= 8:
        hechos.add("defensa_alta")
        hechos.add("defensa_suficiente")

    elif defensa >= 6:
        hechos.add("defensa_suficiente")

    else:
        hechos.add("defensa_baja")


    # FUERZA
    if fuerza >= 8:
        hechos.add("fuerza_alta")
        hechos.add("fuerza_suficiente")

    elif fuerza >= 6:
        hechos.add("fuerza_suficiente")

    else:
        hechos.add("fuerza_baja")


    # REFLEJOS
    if reflejos >= 8:
        hechos.add("reflejos_altos")

    elif reflejos >= 6:
        hechos.add("reflejos_medios")

    else:
        hechos.add("reflejos_bajos")


    # JUEGO AÉREO
    if juego_aereo >= 8:
        hechos.add("juego_aereo_alto")

    elif juego_aereo >= 6:
        hechos.add("juego_aereo_medio")

    else:
        hechos.add("juego_aereo_bajo")


    # Guardamos también los valores originales
    datos_jugador = {
        "nombre": nombre if nombre else "Jugador",
        "velocidad": velocidad,
        "resistencia": resistencia,
        "pase": pase,
        "vision": vision,
        "regate": regate,
        "remate": remate,
        "defensa": defensa,
        "fuerza": fuerza,
        "reflejos": reflejos,
        "juego_aereo": juego_aereo
    }

    return datos_jugador, hechos
```

```python
# ------------------------------------------------------------
# 3. MOTOR DE INFERENCIA
# ------------------------------------------------------------

def motor_inferencia(hechos):
    """
    Compara los hechos detectados del jugador con las condiciones
    de cada regla de la base de conocimientos.

    Si TODAS las condiciones de una regla están presentes,
    la regla se activa y genera una conclusión.
    """

    resultados = []

    print("\n" + "=" * 55)
    print("               MOTOR DE INFERENCIA")
    print("=" * 55)

    # --------------------------------------------------------
    # Mostrar los hechos detectados
    # --------------------------------------------------------

    print("\n Hechos detectados en el jugador:\n")

    for hecho in sorted(hechos):
        print("   •", hecho)

    print("\n Analizando reglas...\n")

    # --------------------------------------------------------
    # Revisamos cada regla de la base de conocimientos
    # --------------------------------------------------------

    for regla in reglas:

        condiciones = regla["condiciones"]

        # Verificamos si TODAS las condiciones
        # de la regla están presentes en los hechos.
        if all(condicion in hechos for condicion in condiciones):

            print(" Regla activada:", regla["nombre"])

            resultados.append(regla)

    return resultados
```

```python
# ------------------------------------------------------------
# 4. MOSTRAR RESULTADOS
# ------------------------------------------------------------

def mostrar_resultados(datos_jugador, resultados):

    print("\n" + "=" * 55)
    print("            RESULTADO DEL SISTEMA ")
    print("=" * 55)

    print(f"\n Jugador evaluado: {datos_jugador['nombre']}")

    # --------------------------------------------------------
    # Si ninguna regla fue activada
    # --------------------------------------------------------

    if not resultados:

        print("\n No se encontró una posición específica")
        print("con las características ingresadas.")

        print("\n Recomendación:")
        print(
            "→ El jugador presenta un perfil equilibrado o todavía "
            "no cumple completamente las condiciones definidas "
            "para una posición específica."
        )

        print(
            "→ Se recomienda seguir entrenando y realizar "
            "una evaluación más detallada."
        )

        return

    # --------------------------------------------------------
    # Mostrar las posiciones encontradas
    # --------------------------------------------------------

    print("\n Posiciones recomendadas:")

    for i, resultado in enumerate(resultados, start=1):

        print("\n" + "-" * 55)

        print(f" OPCIÓN {i}: {resultado['conclusion']}")

        print("\n Regla utilizada:")
        print(f"→ {resultado['nombre']}")

        print("\n Motivo:")
        print(f"→ {resultado['motivo']}")

        print("\n Recomendación de entrenamiento:")
        print(f"→ {resultado['recomendacion']}")

    print("\n" + "-" * 55)

    # --------------------------------------------------------
    # Resumen final
    # --------------------------------------------------------

    if len(resultados) == 1:

        print(
            f"\n Según el sistema experto, la posición más "
            f"compatible con {datos_jugador['nombre']} es:"
        )

        print(f"\n {resultados[0]['conclusion']}")

    else:

        posiciones = [r["conclusion"] for r in resultados]

        print(
            f"\n {datos_jugador['nombre']} presenta características "
            "compatibles con varias posiciones:"
        )

        for posicion in posiciones:
            print(f"    {posicion}")

        print(
            "\n La elección final puede depender de las necesidades "
            "del equipo y del entrenamiento del jugador."
        )
```

```python
# ------------------------------------------------------------
# 5. PROGRAMA PRINCIPAL
# ------------------------------------------------------------

def main():

    # 1. Obtener los datos del jugador
    #    y convertirlos en hechos
    datos_jugador, hechos = obtener_hechos()

    # 2. Ejecutar el motor de inferencia
    resultados = motor_inferencia(hechos)

    # 3. Mostrar las conclusiones
    mostrar_resultados(datos_jugador, resultados)


# Ejecutar el sistema experto
main()
```
